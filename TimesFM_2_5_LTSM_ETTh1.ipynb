## Getting Started - A Simple Example

```python
history = [
    np.linspace(0, 1, 100),
    np.sin(np.linspace(0, 20, 67))
]
```

```python
point_forecast, quantile_forecast = model.forecast(
    horizon = 12,
    inputs = history
)
```

```python
point_forecast.shape, quantile_forecast.shape
```

```python
# plot the history and the forecast including quantiles:
history_df = pd.DataFrame({f'Series_{i+1}': pd.Series(s) for i, s in enumerate(history)})
forecast_df = pd.DataFrame({f'Series_{i+1}': pd.Series(s) for i, s in enumerate(point_forecast)})


fig, axes = plt.subplots(len(history_df.columns), 1, figsize=(10, 3 * len(history_df.columns)))

for i, col in enumerate(history_df.columns):
    # Get the history and drop any NaN values from padding
    history_series = history_df[col].dropna()
    # Get the forecast and drop NaN values
    forecast_series = forecast_df[col].dropna()
    # Set the forecast's index to start after the history ends
    forecast_series.index = range(len(history_series), len(history_series) + len(forecast_series))
    # Plot both series on the same subplot
    history_series.plot(ax=axes[i], label='History')
    forecast_series.plot(ax=axes[i], label='Forecast', style='--')

    # Get the quantiles for the current series (shape will be 12, 10)
    series_quantiles = np.sort(quantile_forecast[i], axis=1)
    axes[i].fill_between(
        forecast_series.index, series_quantiles[:, 1], series_quantiles[:, 8],
        color='skyblue', alpha=0.3, label='80% Interval'
    )
    # Plot the 30th-70th percentile region (narrower, darker)
    axes[i].fill_between(
        forecast_series.index, series_quantiles[:, 3], series_quantiles[:, 6],
        color='skyblue', alpha=0.6, label='40% Interval'
    )

    axes[i].set_title(col)
    axes[i].legend()
    axes[i].grid(True, linestyle=':')

plt.tight_layout()
plt.show()
```

## Time Series Data: BigQuery Public Data

Similar to the other workflows in this repository, [Applied Forecasting](./readme.md), the following pulls a sample of times series from the BigQuery Public Dataset `bigquery-public-data.new_york.citibike_trips`.  A set of bike stations near New York City Central Park.  The time series are for daily demand for bikes trips starting at each station.

```python
ds = bq.query(
"""
SELECT
    start_station_name,
    COUNT(*) as num_trips,
    EXTRACT(DATE FROM starttime) AS starttime
FROM bigquery-public-data.new_york.citibike_trips
WHERE start_station_name LIKE '%Central Park%'
GROUP BY start_station_name, starttime
ORDER BY start_station_name, starttime
"""
).to_dataframe()
```

## TimesFM Forecasting - From Dataframe


```python
history = []
station_names = []
last_dates = ds.groupby('start_station_name')['starttime'].max()
for name, group in ds.groupby('start_station_name'):
    station_names.append(name)
    history.append(group['num_trips'].to_numpy())

HORIZON = 30
point_forecast, quantile_forecast = model.forecast(
    horizon = HORIZON,
    inputs = history
)

forecast_dfs = []
for name, pf, qf in zip(station_names, point_forecast, quantile_forecast):
    # Get the last known date for the current station
    last_date = last_dates[name]

    ## MODIFIED: Generate a date range for the forecast period
    forecast_dates = pd.date_range(
        start=last_date + pd.Timedelta(days=1),
        periods=HORIZON,
        freq='D'
    )

    # Create DataFrames for the forecast values
    quantiles_df = pd.DataFrame(qf, columns=[f'q_{i+1}' for i in range(10)])
    point_df = pd.DataFrame({'point_forecast': pf})

    # Combine them
    station_df = pd.concat([point_df, quantiles_df], axis=1)

    # Add the station name and the new forecast timestamp
    station_df['start_station_name'] = name
    station_df['forecast_timestamp'] = forecast_dates ## MODIFIED

    forecast_dfs.append(station_df)

# Concatenate all individual station DataFrames
final_df = pd.concat(forecast_dfs, ignore_index=True)

# Reorder columns for clarity
final_df = final_df[['start_station_name', 'forecast_timestamp', 'point_forecast'] + [f'q_{i+1}' for i in range(10)]]

# Display the final, combined result
print("Final Combined DataFrame with Timestamps:")
print(final_df.head())
```

### Visualize Time Series Wih Forecast And Quantiles

```python
# Get the unique list of series to plot
unique_stations = ds['start_station_name'].unique()

# Create a grid with 2 columns
fig, axes = plt.subplots(
    len(unique_stations),
    2, # Two columns for split view
    figsize=(16, 5 * len(unique_stations))
)
if len(unique_stations) == 1: axes = np.array([axes]) # Ensure axes is always 2D

for i, name in enumerate(unique_stations):
    ax_full = axes[i, 0]
    ax_zoom = axes[i, 1]

    # Filter data for the current station
    history_data = ds[ds['start_station_name'] == name]
    forecast_data = final_df[final_df['start_station_name'] == name]

    # --- Plot on the LEFT (Full View) ---
    ax_full.plot(history_data['starttime'], history_data['num_trips'], label='History', color='steelblue')
    ax_full.plot(forecast_data['forecast_timestamp'], forecast_data['point_forecast'], label='Forecast', color='darkorange', linestyle='--')
    ax_full.fill_between(forecast_data['forecast_timestamp'], forecast_data['q_2'], forecast_data['q_9'], color='sandybrown', alpha=0.4, label='80% Interval')
    ax_full.set_title(f'Full History & Forecast: {name}')
    ax_full.legend()
    ax_full.grid(True, linestyle=':')

    # --- Plot on the RIGHT (Zoomed View) ---
    ax_zoom.plot(history_data['starttime'], history_data['num_trips'], label='History', color='steelblue')
    ax_zoom.plot(forecast_data['forecast_timestamp'], forecast_data['point_forecast'], label='Forecast', color='darkorange', linestyle='--')
    ax_zoom.fill_between(forecast_data['forecast_timestamp'], forecast_data['q_2'], forecast_data['q_9'], color='sandybrown', alpha=0.4)
    ax_zoom.set_title(f'Zoomed Forecast: {name}')
    ax_zoom.grid(True, linestyle=':')

    # KEY STEP: Set the x-axis limits for the zoom
    zoom_start = history_data['starttime'].max() - pd.Timedelta(days=30)
    zoom_end = forecast_data['forecast_timestamp'].max() + pd.Timedelta(days=1)
    ax_zoom.set_xlim(zoom_start, zoom_end)

plt.tight_layout()
plt.show()
```
