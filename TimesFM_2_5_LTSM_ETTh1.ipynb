```python
# ============================================================
# TimesFM 2.5 — ETTh1 Univariate Forecasting
# Google Colab — Single Cell
# ============================================================

# 1. Install official TimesFM 2.5 implementation
!git clone -q https://github.com/google-research/timesfm.git /content/timesfm
!pip install -q -e "/content/timesfm[torch]"
```

```python
# 2. Imports
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch

import timesfm


# ============================================================
# Configuration
# ============================================================

DATA_URL = (
    "https://raw.githubusercontent.com/zhouhaoyi/"
    "ETDataset/main/ETT-small/ETTh1.csv"
)

TARGET = "OT"
PREDICTION_LENGTH = 48

# Use this to make the experiment directly comparable
# with your Chronos experiments.
CONTEXT_LENGTH = 512

MODEL_ID = "google/timesfm-2.5-200m-pytorch"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Device: {DEVICE}")
print(f"Target: {TARGET}")
print(f"Prediction length: {PREDICTION_LENGTH}")
print(f"Context length: {CONTEXT_LENGTH}")
```

```python
# ============================================================
# 3. Load ETTh1
# ============================================================

df = pd.read_csv(DATA_URL)

print("\nDataset:")
print(df.shape)
print(df.columns.tolist())

# Convert target to numeric
series = pd.to_numeric(df[TARGET], errors="coerce")

if series.isna().any():
    raise ValueError(
        f"{TARGET} contains {series.isna().sum()} missing values."
    )

full_series = series.to_numpy(dtype=np.float32)

print(f"\nFull history length: {len(full_series)}")
print(f"Target range: {full_series.min():.4f} → {full_series.max():.4f}")
```

```python
# ============================================================
# 4. Train / Test split
#    Last 48 observations = test horizon
# ============================================================

train = full_series[:-PREDICTION_LENGTH]
actual = full_series[-PREDICTION_LENGTH:]

if CONTEXT_LENGTH > len(train):
    raise ValueError(
        f"CONTEXT_LENGTH={CONTEXT_LENGTH} exceeds "
        f"available training history={len(train)}."
    )

history = train[-CONTEXT_LENGTH:]

print(f"\nTraining history: {len(train)}")
print(f"Context used:     {len(history)}")
print(f"Test horizon:     {len(actual)}")
```

```python
# ============================================================
# 5. Load TimesFM 2.5
# ============================================================

print("\nLoading TimesFM 2.5...")

model = timesfm.TimesFM_2p5_200M_torch.from_pretrained(
    MODEL_ID
)

model.compile(
    timesfm.ForecastConfig(
        max_context=CONTEXT_LENGTH,
        max_horizon=PREDICTION_LENGTH,
        normalize_inputs=True,
        use_continuous_quantile_head=True,
        force_flip_invariance=True,
        infer_is_positive=False,
        fix_quantile_crossing=True,
    )
)

print("Model loaded and compiled.")
```

```python
# ============================================================
# 6. Forecast
# ============================================================

print("\nRunning TimesFM 2.5 forecast...")

start_time = time.perf_counter()

point_forecast, quantile_forecast = model.forecast(
    horizon=PREDICTION_LENGTH,
    inputs=[history],
)

inference_time = time.perf_counter() - start_time

prediction = point_forecast[0]

print(f"Inference time: {inference_time:.3f} seconds")
print(f"Forecast shape: {prediction.shape}")
```

```python
# ============================================================
# TimesFM 2.5 Custom Model for AutoGluon 1.6.1
#
# IMPORTANT:
# The compiled TimesFM object is NEVER persisted inside the
# AutoGluon model.
#
# AutoGluon stores only configuration.
# TimesFM is recreated + compiled when prediction is needed.
# ============================================================

import numpy as np
import pandas as pd
import torch
import timesfm

from autogluon.timeseries.models.abstract import (
    AbstractTimeSeriesModel
)

from autogluon.timeseries import (
    TimeSeriesDataFrame
)


class TimesFM25Model(AbstractTimeSeriesModel):

    _supports_known_covariates = False
    _supports_past_covariates = False
    _supports_static_features = False

    # --------------------------------------------------------
    # Constructor
    # --------------------------------------------------------

    def __init__(
        self,
        freq=None,
        prediction_length=1,
        path=None,
        name=None,
        eval_metric=None,
        hyperparameters=None,
        **kwargs,
    ):

        if hyperparameters is None:
            hyperparameters = {}

        hyperparameters = dict(hyperparameters)

        self.context_length = hyperparameters.get(
            "context_length",
            512,
        )

        self.device = hyperparameters.get(
            "device",
            "cuda" if torch.cuda.is_available() else "cpu",
        )

        self.model_id = hyperparameters.get(
            "model_id",
            "google/timesfm-2.5-200m-pytorch",
        )

        # ----------------------------------------------------
        # Runtime-only object
        #
        # NEVER rely on this being serialized.
        # ----------------------------------------------------

        self.model = None

        super().__init__(
            freq=freq,
            prediction_length=prediction_length,
            path=path,
            name=name,
            eval_metric=eval_metric,
            hyperparameters=hyperparameters,
            **kwargs,
        )

    # ========================================================
    # Create TimesFM runtime model
    # ========================================================

    def _load_model(self):

        if self.model is not None:
            return

        print(
            f"Loading TimesFM 2.5: {self.model_id}"
        )

        # ----------------------------------------------------
        # Load pretrained model
        # ----------------------------------------------------

        model = (
            timesfm.TimesFM_2p5_200M_torch
            .from_pretrained(
                self.model_id
            )
        )

        print(
            "TimesFM 2.5 loaded successfully."
        )

        print(
            "Model type:",
            type(model)
        )

        # ----------------------------------------------------
        # Forecast configuration
        # ----------------------------------------------------

        forecast_config = (
            timesfm.ForecastConfig(
                max_context=self.context_length,
                max_horizon=self.prediction_length,
                normalize_inputs=True,
                use_continuous_quantile_head=True,
                force_flip_invariance=True,
                infer_is_positive=False,
                fix_quantile_crossing=True,
            )
        )

        # ----------------------------------------------------
        # Compile
        # ----------------------------------------------------

        print(
            "Compiling TimesFM 2.5..."
        )

        model.compile(
            forecast_config
        )

        print(
            "TimesFM 2.5 compiled successfully."
        )

        # ----------------------------------------------------
        # Only now attach to runtime object.
        #
        # The object will be removed before AutoGluon saves.
        # ----------------------------------------------------

        self.model = model

    # ========================================================
    # Remove runtime TimesFM before AutoGluon serialization
    # ========================================================

    def _clear_runtime_model(self):

        if self.model is not None:

            print(
                "Removing compiled TimesFM "
                "runtime object before serialization."
            )

        self.model = None

    # ========================================================
    # Custom save
    # ========================================================
    #
    # This is the important part.
    #
    # AutoGluon calls model.save().
    #
    # We temporarily remove TimesFM, let the parent class
    # pickle the model, then restore it.
    #
    # ========================================================

    def save(
        self,
        path=None,
        verbose=True,
    ):

        runtime_model = self.model

        # Remove compiled TimesFM
        self.model = None

        try:

            return super().save(
                path=path,
                verbose=verbose,
            )

        finally:

            # Restore runtime model for current process
            self.model = runtime_model

    # ========================================================
    # Fit
    # ========================================================

    def _fit(
        self,
        train_data,
        val_data=None,
        time_limit=None,
        **kwargs,
    ):

        print(
            "\nTimesFM25Model._fit()"
        )

        print(
            "Training data:",
            train_data.shape
        )

        print(
            "Prediction length:",
            self.prediction_length
        )

        print(
            "Context length:",
            self.context_length
        )

        # ----------------------------------------------------
        # Load + compile for AutoGluon validation.
        # ----------------------------------------------------

        self._load_model()

        return self

    # ========================================================
    # Predict
    # ========================================================

    def _predict(
        self,
        data,
        known_covariates=None,
        **kwargs,
    ):

        print(
            "\nTimesFM25Model._predict()"
        )

        # ----------------------------------------------------
        # Important:
        #
        # If the model was loaded from disk, self.model will
        # be None because we deliberately don't serialize it.
        #
        # Therefore _load_model() recreates it.
        # ----------------------------------------------------

        self._load_model()

        # ----------------------------------------------------
        # Determine target column
        # ----------------------------------------------------

        print(
            "Prediction data columns:",
            list(data.columns)
        )

        target_column = getattr(
            self,
            "target",
            None,
        )

        if (
            target_column is not None
            and target_column in data.columns
        ):
            pass

        elif "target" in data.columns:

            target_column = "target"

        elif len(data.columns) == 1:

            target_column = data.columns[0]

        else:

            raise ValueError(
                "Could not determine target column.\n"
                f"Available columns: "
                f"{list(data.columns)}\n"
                f"Configured target: "
                f"{getattr(self, 'target', None)}"
            )

        print(
            "Using target column:",
            target_column
        )

        # ----------------------------------------------------
        # Get series
        # ----------------------------------------------------

        item_ids = data.item_ids

        predictions = []

        # ----------------------------------------------------
        # Forecast each series
        # ----------------------------------------------------

        for item_id in item_ids:

            item_data = data.loc[item_id]

            series = (
                item_data[target_column]
                .to_numpy(
                    dtype=np.float32
                )
            )

            # Most recent context
            context = series[
                -self.context_length:
            ]

            if len(context) < self.context_length:

                raise ValueError(
                    f"Series {item_id} has "
                    f"{len(context)} observations, "
                    f"but context_length="
                    f"{self.context_length}"
                )

            print(
                f"  {item_id}: "
                f"context={len(context)}, "
                f"horizon={self.prediction_length}"
            )

            # ------------------------------------------------
            # TimesFM input
            # ------------------------------------------------

            inputs = [
                context
            ]

            # ------------------------------------------------
            # Forecast
            # ------------------------------------------------

            result = self.model.forecast(
                horizon=self.prediction_length,
                inputs=inputs,
            )

            forecast = np.asarray(
                result[0]
            )

            print(
                "    Raw forecast shape:",
                forecast.shape
            )

            # Remove batch dimension
            if forecast.ndim == 2:

                forecast = forecast[0]

            elif forecast.ndim != 1:

                raise ValueError(
                    "Unexpected TimesFM forecast "
                    f"shape: {forecast.shape}"
                )

            forecast = forecast[
                :self.prediction_length
            ]

            if len(forecast) != self.prediction_length:

                raise ValueError(
                    f"TimesFM returned "
                    f"{len(forecast)} values, "
                    f"expected "
                    f"{self.prediction_length}"
                )

            predictions.append(
                forecast
            )

        # ----------------------------------------------------
        # Numpy
        # ----------------------------------------------------

        predictions = np.asarray(
            predictions,
            dtype=np.float32,
        )

        print(
            "Final predictions shape:",
            predictions.shape
        )

        # ----------------------------------------------------
        # AutoGluon output
        # ----------------------------------------------------

        rows = []

        for i, item_id in enumerate(item_ids):

            item_data = data.loc[item_id]

            last_timestamp = (
                item_data.index[-1]
            )

            future_timestamps = pd.date_range(
                start=last_timestamp,
                periods=self.prediction_length + 1,
                freq=self.freq,
            )[1:]

            for j, timestamp in enumerate(
                future_timestamps
            ):

                rows.append(
                    {
                        "item_id": item_id,
                        "timestamp": timestamp,
                        "mean": float(
                            predictions[i, j]
                        ),
                    }
                )

        prediction_df = pd.DataFrame(
            rows
        )

        return TimeSeriesDataFrame(
            prediction_df
        )
```
